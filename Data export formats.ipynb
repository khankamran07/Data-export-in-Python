```python
import pandas as pd
df = pd.DataFrame()
```

## Export to CSV 

```python
df = pd.read_csv('deliveries.csv')
df.head()
```

```python
#  Suppose someone asks you to create a new dataset from this dataset,
# where the new column should contain the batsman's name and  the number of runs scored by each batsman, and then export this DataFrame as a CSV file.


temp_df = df.groupby('batsman')['batsman_runs'].sum()
```

```python
temp_df.to_csv('batsman_runs.csv')
```

```python
# jab aap csv file kholege to agar extra col nhi chaiye to aap likhoge index=False

temp_df.to_csv('batsman_runs.csv',index=False)
```

```python
df.pivot_table(index = 'batsman',columns = 'bowling_team',values = 'batsman_runs',aggfunc = 'sum').to_csv("batsman_vs_team.csv")
```

## Export to Excel

```python
df
```

```python
temp_df = df.groupby('batsman')['batsman_runs'].sum().reset_index()
temp_df
```

```python
# Change the above temp_df into excel

temp_df.to_excel('batsman_runs.xlsx')
```

```python
# If you want, you can also store it in a particular sheet.

temp_df.to_excel('output.xlsx',sheet_name = 'batsman_runs')
```

```python
# Suppose you have to create an Excel file that contains multiple sheets.
# What we are doing is storing the batsman's runs in one sheet and the pivot table in another sheet.


temp_df2 = df.pivot_table(index = 'batsman',columns = 'bowling_team',values = 'batsman_runs',aggfunc = 'sum')
```

So, what you have to do below? You have to write the code in such a way that whenever you have an Excel file and you want to work with multiple sheets in that file, you use an Excel Writer object.

First, you create an object of the Excel Writer class, and then you provide the name of your file. We can call it `output.xlsx` as `writer`.

After that, you take all your DataFrames and write them one by one. For example, if my first DataFrame is `temp_df`, I call `temp_df.to_excel()` and pass the writer object and specify the sheet name.

Then, for my second DataFrame, I do the same thing: `temp_df2.to_excel()`, pass the writer object, and specify another sheet name.

So basically, what am I doing? I am creating a **single Excel file**, but I am working with **multiple sheets** inside that same file. That's why, when I run the code, it creates one Excel file containing both sheets.


```python
with pd.ExcelWriter('output.xlsx') as writer:
    temp_df.to_excel(writer,sheet_name = 'Sheet_name_1')
    temp_df2.to_excel(writer,sheet_name = 'Sheet_name_2')
    
```

## 4.Export to HTML files

```python
# You can convert your Pandas DataFrame into HTML.
# Let's say you want to write your analysis in the form of a blog, and you want to display your analysis as a table in the blog.
```

```python
# We are creating a DataFrame using `df.pivot_table`. This pivot table contains information related to overs and balls. 

df.query('batsman_runs == 6').pivot_table(index='over',columns='ball',values='batsman_runs',aggfunc='count').to_html('sixes_heatmap.html')
```

```python
# From the `df` above, you have to create a JSON. In that JSON, 
# there will be the name of each batsman, the name of the team they played for, and the number of runs they scored.
 

df.groupby(['batting_team','batsman'])['batsman_runs'].sum().unstack().to_json('ipl.json')
```
